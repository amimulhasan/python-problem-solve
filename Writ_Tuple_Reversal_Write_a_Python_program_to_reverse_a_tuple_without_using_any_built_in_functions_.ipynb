```python
# Tuple Reversal: Write a Python program to reverse a tuple without using any built-in functions.
# tupl=(2,4,5,6,7,8,1,9)
def reverse_tuple(tupl):
  reversed_tuple=()
  for i in range(len(tupl)-1,-1,-1):
    reversed_tuple +=(tupl[i],)
  return reversed_tuple

tupl=(2,4,5,6,7,8,1,9)
new_tuple=reverse_tuple(tupl)
print(new_tuple)
```
