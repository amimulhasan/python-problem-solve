```python
# Write a Python program that takes a set as input and prints the number of elements in the set.
# Take input from the user as a comma-separated string
input_str = input("Enter the elements of the set, separated by commas: ")

# Convert the input string to a set
user_set = set(input_str.split(','))

# Remove any leading/trailing whitespace from elements
user_set = {elem.strip() for elem in user_set}

# Print the set and its size
print("Your set:", user_set)
print("Number of elements in the set:", len(user_set))
```
