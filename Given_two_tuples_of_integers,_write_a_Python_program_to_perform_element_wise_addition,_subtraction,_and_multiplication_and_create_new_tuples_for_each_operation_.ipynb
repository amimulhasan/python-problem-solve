```python
#  Given two tuples of integers, write a Python program to perform element-wise addition, subtraction, and multiplication and create new tuples for each operation.
# a=(1,2,3,4,5,6)
# b=(6,7,8,9,4,5)
def element_wise_operation(a,b):
  if len(a) != len(b):
    raise ValueError('Tuple must be same of the length')
  # Perform element-wise operations
  addition=tuple(a+b for a,b in zip(a,b))
  subtraction=tuple(a-b for a,b in zip(a,b))
  multiplication=tuple(a*b for a,b in zip(a,b))
  return addition,subtraction,multiplication
a=(1,2,3,4,5,6)
b=(6,7,8,9,4,5)
addition,subtraction,multiplication=element_wise_operation(a,b)
print('Addition:',addition)
print('Subtraction:',subtraction)
print('Multiplication:',multiplication)
```
