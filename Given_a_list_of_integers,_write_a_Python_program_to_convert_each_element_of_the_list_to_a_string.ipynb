```python
# List of integers
numbers = [1, 2, 3, 4, 5, 6, 7]

# Convert each integer to a string using list comprehension
string_numbers = [str(num) for num in numbers]

# Display the result
print("Original list:", numbers)
print("List with strings:", string_numbers)
```
