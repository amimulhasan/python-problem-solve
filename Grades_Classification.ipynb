```python
def get_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    else:
        return "Fail"

# Taking input from the user
try:
    percentage = float(input("Enter student percentage: "))
    if 0 <= percentage <= 100:
        print(f"Grade: {get_grade(percentage)}")
    else:
        print("Error: Please enter a percentage between 0 and 100.")
except ValueError:
    print("Error: Invalid input! Please enter a numeric value.")
```
