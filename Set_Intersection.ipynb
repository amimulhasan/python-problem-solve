```python

#  Given two sets A and B, write a Python program to find their intersection and print the common elements.


# Define two sets
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

# Find intersection
common_elements = A.intersection(B)

# Print the result
print("Common elements:", common_elements)
```
