```python
# Given a tuple containing various elements, write a Python program to count the frequency of a specific element in the tuple.
def count_element_frequency(tup,element):
  return tup.count(element)

tup=(1,2,3,4,5,6,78,7,5,3,9,5,55)
element=5
fre=count_element_frequency(tup,element)
print(f'The element {element} appears {fre} times in the tuple.')
```
