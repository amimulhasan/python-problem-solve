```python
#  Write a Python program that takes an element as input and checks if it exists in a given set. Print “Found” if the element is present and “Not Found” otherwise.

# Define a sample set
sample_set = {10, 20, 30, 40, 50}

# Take input from the user
element = input("Enter an element to search: ")

# Try to convert input to the appropriate type (int)
try:
    element = int(element)
except ValueError:
    print("Invalid input. Please enter an integer.")
    exit()

# Check if the element exists in the set
if element in sample_set:
    print("Found")
else:
    print("Not Found")

```
