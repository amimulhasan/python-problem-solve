```python
def greet_based_on_time(hour):
    try:
        hour = int(hour)
        if hour < 0 or hour > 23:
            return "Invalid input! Please enter an hour between 0 and 23."

        if 5 <= hour < 12:
            return "Good Morning"
        elif 12 <= hour < 17:
            return "Good Afternoon"
        elif 17 <= hour < 21:
            return "Good Evening"
        else:
            return "Good Night"
    except ValueError:
        return "Invalid input! Please enter a valid integer for the hour."

# Taking input from the user
time_input = input("Enter the time in hours (0-23): ")
greeting = greet_based_on_time(time_input)
print(greeting)
```
