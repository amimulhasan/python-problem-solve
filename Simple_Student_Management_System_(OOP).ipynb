```python
class Student:
    def __init__(self, student_id, name, department, gpa):
        self.student_id = student_id
        self.name = name
        self.department = department
        self.gpa = gpa

    def display_info(self):
        print(f"ID: {self.student_id}, Name: {self.name}, Dept: {self.department}, GPA: {self.gpa}")


class StudentManager:
    def __init__(self):
        self.students = []

    def add_student(self, student):
        self.students.append(student)
        print("✅ Student added successfully.")

    def display_all_students(self):
        if not self.students:
            print("❌ No students to display.")
        else:
            print("📄 Student List:")
            for student in self.students:
                student.display_info()

    def search_student(self, student_id):
        for student in self.students:
            if student.student_id == student_id:
                print("🔍 Student Found:")
                student.display_info()
                return
        print("❌ Student not found.")


# Main program
manager = StudentManager()

while True:
    print("\n===== Student Management System =====")
    print("1. Add Student")
    print("2. View All Students")
    print("3. Search Student by ID")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == '1':
        student_id = input("Enter Student ID: ")
        name = input("Enter Name: ")
        department = input("Enter Department: ")
        gpa = float(input("Enter GPA: "))
        student = Student(student_id, name, department, gpa)
        manager.add_student(student)

    elif choice == '2':
        manager.display_all_students()

    elif choice == '3':
        search_id = input("Enter Student ID to search: ")
        manager.search_student(search_id)

    elif choice == '4':
        print("👋 Exiting... Goodbye!")
        break

    else:
        print("⚠️ Invalid choice. Try again.")
```
