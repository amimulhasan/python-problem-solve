```python
# Write a Python program that takes a list of elements as input and creates a new set containing only the unique elements from the list.
# Take input from the user
input_list = input("Enter elements separated by spaces: ").split()

# Create a set from the list to remove duplicates
unique_elements = set(input_list)

# Print the set of unique elements
print("Unique elements as a set:", unique_elements)
```

```python
# Take input from the user and convert each element to an integer
input_list = list(map(int, input("Enter numbers separated by spaces: ").split()))

# Create a set from the list to remove duplicates
unique_elements = set(input_list)

# Print the set of unique elements
print("Unique numbers as a set:", unique_elements)
```
