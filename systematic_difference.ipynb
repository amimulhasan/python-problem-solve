```python


# Given two sets A and B, write a Python program to find the symmetric difference between the two sets (i.e., elements that are present in either set A or set B, but not in both) and print the result.


# Define two sets A and B
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

# Find the symmetric difference using ^ operator or symmetric_difference() method
symmetric_diff = A ^ B  # You can also use A.symmetric_difference(B)

# Print the result
print("Symmetric Difference between A and B:", symmetric_diff)
```
