```python
'''Dictionary Value Search: Given a dictionary of items and their prices,
write a Python program to search for an item based on its price and print the item’s name. '''


# Dictionary of items and their prices
items = {
    "apple": 100,
    "banana": 40,
    "orange": 80,
    "grapes": 120,
    "watermelon": 150
}

# Function to search for an item by price
def search_item_by_price(price):
    found = False
    for item, item_price in items.items():
        if item_price == price:
            print(f"Item found: {item} (Price: {price})")
            found = True
    if not found:
        print(f"No item found with price {price}")

# Example: Search for an item with price 80
search_price = int(input("Enter the price to search: "))
search_item_by_price(search_price)
```
