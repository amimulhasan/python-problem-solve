```python
# Dictionary Merging: Given two dictionaries, write a Python program to merge them into a single dictionary and print the result.


dic_1={'rofiq':1,'sakib':2,'fahad':4}
dic_2={'a':5,'b':8,'c':9}
marg_dic={**dic_1,**dic_2}
print(marg_dic)
```
