```python
# Write a Python program that takes an integer as input and prints whether the number falls within the ranges: 0-50, 51-100, 101-150, or above 150.

def check_number_range(num):
    if 0 <= num <= 50:
        print(f"{num} falls within the range 0-50.")
    elif 51 <= num <= 100:
        print(f"{num} falls within the range 51-100.")
    elif 101 <= num <= 150:
        print(f"{num} falls within the range 101-150.")
    elif num > 150:
        print(f"{num} is above 150.")
    else:
        print(f"{num} is a negative number.")

try:
    number = int(input("Enter an integer: "))
    check_number_range(number)
except ValueError:
    print("Invalid input! Please enter a valid integer.")

```
