```python
# Dictionary Length: Write a Python program to calculate and print the number of key-value pairs in a given dictionary.




# Sample dictionary
my_dict = {
    "name": "Rofiq",
    "age": 22,
    "department": "Biomedical Engineering",
    "university": "Islamic University"
}

# Calculate the number of key-value pairs
length = len(my_dict)

# Print the result
print("Number of key-value pairs in the dictionary:", length)
```
