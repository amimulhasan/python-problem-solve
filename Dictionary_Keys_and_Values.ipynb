```python
# Dictionary Keys and Values: Write a Python program that takes a dictionary as input and prints all the keys and values in separate lines.
# Sample dictionary (you can modify or input your own)
my_dict = {
    "name": "Rofiq",
    "department": "Biomedical Engineering",
    "university": "Islamic University",
    "year": 3
}

# Printing keys
print("Keys:")
for key in my_dict.keys():
    print(key)

# Printing values
print("\nValues:")
for value in my_dict.values():
    print(value)

```
