```python
# Existing dictionary of students and their scores
''' 77. Dictionary Manipulation: Given a dictionary with student names as keys and their corresponding scores as values, write a Python
 program to add a new student to the dictionary and update the score of an existing student.'''


student_scores = {
    "Alice": 85,
    "Bob": 90,
    "Charlie": 78
}

# Add a new student
new_student = "David"
new_score = 88
student_scores[new_student] = new_score
print(f"Added new student: {new_student} with score {new_score}")

# Update the score of an existing student
existing_student = "Bob"
updated_score = 95
student_scores[existing_student] = updated_score
print(f"Updated {existing_student}'s score to {updated_score}")

# Print the final dictionary
print("\nFinal student scores:")
for name, score in student_scores.items():
    print(f"{name}: {score}")
```
